## Question 41: Student Performance System

```python
# Given data
students = [
    {"id": 1, "name": "Ali", "marks": (85, 90, 88)},
    {"id": 2, "name": "Sara", "marks": (92, 95, 91)},
    {"id": 3, "name": "Ahmed", "marks": (78, 84, 80)}
]

# Calculate each student's average without using loops.
averages = {
    students[0]["name"]: sum(students[0]["marks"]) / len(students[0]["marks"]),
    students[1]["name"]: sum(students[1]["marks"]) / len(students[1]["marks"]),
    students[2]["name"]: sum(students[2]["marks"]) / len(students[2]["marks"])
}
top_student = max(averages, key=averages.get)
student_names = (students[0]["name"], students[1]["name"], students[2]["name"])
unique_marks = set(students[0]["marks"]) | set(students[1]["marks"]) | set(students[2]["marks"])

print("Student averages:", averages)
print("Top student:", top_student)
print("Tuple of names:", student_names)
print("Unique marks:", unique_marks)
```

## Question 42: Employee Skills

```python
# Given data
employees = {
    "Ali": {"Python", "SQL", "AWS"},
    "Sara": {"Python", "Databricks", "SQL"},
    "Ahmed": {"AWS", "Snowflake", "SQL"}
}

# Use set operations to compare employee skills.
common_skills = employees["Ali"] & employees["Sara"] & employees["Ahmed"]
ali_only_skills = employees["Ali"] - employees["Sara"] - employees["Ahmed"]
all_team_skills = employees["Ali"] | employees["Sara"] | employees["Ahmed"]
python_sql_employees = (
    ({"Ali"} if {"Python", "SQL"} <= employees["Ali"] else set())
    | ({"Sara"} if {"Python", "SQL"} <= employees["Sara"] else set())
    | ({"Ahmed"} if {"Python", "SQL"} <= employees["Ahmed"] else set())
)

print("Common skills:", common_skills)
print("Ali-only skills:", ali_only_skills)
print("All team skills:", all_team_skills)
print("Employees with Python and SQL:", python_sql_employees)
```

## Question 43: Sales Data

```python
# Given data
sales = [
    ("Laptop", 120000, 2),
    ("Mouse", 1500, 5),
    ("Keyboard", 3500, 3)
]

# Calculate revenue per product from price times quantity.
revenues = {
    sales[0][0]: sales[0][1] * sales[0][2],
    sales[1][0]: sales[1][1] * sales[1][2],
    sales[2][0]: sales[2][1] * sales[2][2]
}
total_revenue = sum(revenues.values())
top_revenue_product = max(revenues, key=revenues.get)
name_to_revenue = revenues.copy()
prices = {sales[0][1], sales[1][1], sales[2][1]}

print("Revenue by product:", revenues)
print("Total revenue:", total_revenue)
print("Top-revenue product:", top_revenue_product)
print("Name-to-revenue dictionary:", name_to_revenue)
print("Set of prices:", prices)
```

## Question 44: Customer Orders

```python
# Given data
customers = {
    "C101": {"name": "Ali", "orders": [1200, 2500, 1800]},
    "C102": {"name": "Sara", "orders": [3000, 4500]},
    "C103": {"name": "Ahmed", "orders": [1500, 2000, 1000]}
}

# Calculate each customer's spending.
spending = {
    "C101": sum(customers["C101"]["orders"]),
    "C102": sum(customers["C102"]["orders"]),
    "C103": sum(customers["C103"]["orders"])
}
top_customer_id = max(spending, key=spending.get)
top_customer = customers[top_customer_id]["name"]
totals_in_id_order = (spending["C101"], spending["C102"], spending["C103"])

print("Spending per customer:", spending)
print("Top customer:", top_customer)
print("Tuple of totals in customer ID order:", totals_in_id_order)
```

## Question 45: Course Enrollment

```python
# Given data
courses = {
    "Python": {"Ali", "Sara", "Ahmed"},
    "SQL": {"Sara", "Ahmed", "Zain"},
    "AWS": {"Ali", "Ahmed", "Zain"}
}

# Build groups using set operations.
python_sql = courses["Python"] & courses["SQL"]
python_aws = courses["Python"] & courses["AWS"]
sql_aws = courses["SQL"] & courses["AWS"]
in_exactly_two = (python_sql | python_aws | sql_aws) - (courses["Python"] & courses["SQL"] & courses["AWS"])
in_all_three = courses["Python"] & courses["SQL"] & courses["AWS"]
in_exactly_one = (
    (courses["Python"] - courses["SQL"] - courses["AWS"])
    | (courses["SQL"] - courses["Python"] - courses["AWS"])
    | (courses["AWS"] - courses["Python"] - courses["SQL"])
)
all_students = courses["Python"] | courses["SQL"] | courses["AWS"]

print("Students in exactly two courses:", in_exactly_two)
print("Students in all three courses:", in_all_three)
print("Students in exactly one course:", in_exactly_one)
print("Number of unique students:", len(all_students))
```

## Question 46: E-Commerce Transactions

```python
# Given data
transactions = [
    {"id": 101, "customer": "Ali", "items": [("Laptop", 120000, 1), ("Mouse", 1500, 2)]},
    {"id": 102, "customer": "Sara", "items": [("Keyboard", 3500, 2), ("Mouse", 1500, 1)]},
    {"id": 103, "customer": "Ali", "items": [("Monitor", 45000, 1)]}
]

# Calculate each transaction total from its items.
transaction_totals = {
    transactions[0]["id"]: (
        transactions[0]["items"][0][1] * transactions[0]["items"][0][2]
        + transactions[0]["items"][1][1] * transactions[0]["items"][1][2]
    ),
    transactions[1]["id"]: (
        transactions[1]["items"][0][1] * transactions[1]["items"][0][2]
        + transactions[1]["items"][1][1] * transactions[1]["items"][1][2]
    ),
    transactions[2]["id"]: (
        transactions[2]["items"][0][1] * transactions[2]["items"][0][2]
    )
}
total_revenue = sum(transaction_totals.values())
ali_spending = transaction_totals[101] + transaction_totals[103]
largest_transaction_id = max(transaction_totals, key=transaction_totals.get)
unique_products = (
    {transactions[0]["items"][0][0], transactions[0]["items"][1][0]}
    | {transactions[1]["items"][0][0], transactions[1]["items"][1][0]}
    | {transactions[2]["items"][0][0]}
)

print("Total revenue:", total_revenue)
print("Ali spending:", ali_spending)
print("Largest transaction ID:", largest_transaction_id)
print("Largest transaction amount:", transaction_totals[largest_transaction_id])
print("Unique product names:", unique_products)
```

## Question 47: Pipeline Monitoring

```python
# Given data
pipelines = {
    "pipeline_1": {"status": "success", "records": (1000, 1200, 1100)},
    "pipeline_2": {"status": "failed", "records": (800, 0, 0)},
    "pipeline_3": {"status": "success", "records": (1500, 1600, 1700)}
}

# Calculate totals and averages for each pipeline.
pipeline_totals = {
    "pipeline_1": sum(pipelines["pipeline_1"]["records"]),
    "pipeline_2": sum(pipelines["pipeline_2"]["records"]),
    "pipeline_3": sum(pipelines["pipeline_3"]["records"])
}
total_processed_records = sum(pipeline_totals.values())
failed_pipelines = (
    {"pipeline_1"} if pipelines["pipeline_1"]["status"] == "failed" else set()
) | (
    {"pipeline_2"} if pipelines["pipeline_2"]["status"] == "failed" else set()
) | (
    {"pipeline_3"} if pipelines["pipeline_3"]["status"] == "failed" else set()
)
largest_pipeline = max(pipeline_totals, key=pipeline_totals.get)
average_records_per_run = {
    "pipeline_1": sum(pipelines["pipeline_1"]["records"]) / len(pipelines["pipeline_1"]["records"]),
    "pipeline_2": sum(pipelines["pipeline_2"]["records"]) / len(pipelines["pipeline_2"]["records"]),
    "pipeline_3": sum(pipelines["pipeline_3"]["records"]) / len(pipelines["pipeline_3"]["records"])
}

print("Total processed records:", total_processed_records)
print("Failed pipelines:", failed_pipelines)
print("Largest pipeline by total:", largest_pipeline)
print("Pipeline totals:", pipeline_totals)
print("Average records per run:", average_records_per_run)
```

## Question 48: Banking Transactions

```python
# Given data
accounts = {
    "A101": {"holder": "Ali", "transactions": [5000, -2000, 3000, -1000]},
    "A102": {"holder": "Sara", "transactions": [10000, -3000, -2000]},
    "A103": {"holder": "Ahmed", "transactions": [7000, -1500, 2500]}
}

# Starting from zero, the final balance is the sum of each account's transactions.
final_balances = {
    "A101": sum(accounts["A101"]["transactions"]),
    "A102": sum(accounts["A102"]["transactions"]),
    "A103": sum(accounts["A103"]["transactions"])
}
highest_balance_account = max(final_balances, key=final_balances.get)
total_withdrawals = (
    abs(min(accounts["A101"]["transactions"][1], 0))
    + abs(min(accounts["A101"]["transactions"][3], 0))
    + abs(min(accounts["A102"]["transactions"][1], 0))
    + abs(min(accounts["A102"]["transactions"][2], 0))
    + abs(min(accounts["A103"]["transactions"][1], 0))
)
largest_deposit = max(
    max(accounts["A101"]["transactions"]),
    max(accounts["A102"]["transactions"]),
    max(accounts["A103"]["transactions"])
)

print("Final balances:", final_balances)
print("Highest balance:", accounts[highest_balance_account]["holder"], final_balances[highest_balance_account])
print("Total withdrawals:", total_withdrawals)
print("Largest single deposit:", largest_deposit)
```

## Question 49: Product Recommendations

```python
# Given data
users = {
    "Ali": {"purchased": {"Laptop", "Mouse"}, "wishlist": {"Keyboard", "Monitor", "Mouse"}},
    "Sara": {"purchased": {"Keyboard", "Monitor"}, "wishlist": {"Laptop", "Mouse"}},
    "Ahmed": {"purchased": {"Laptop", "Keyboard"}, "wishlist": {"Monitor", "Headphones"}}
}

# Products purchased by at least two users.
ali_purchases = users["Ali"]["purchased"]
sara_purchases = users["Sara"]["purchased"]
ahmed_purchases = users["Ahmed"]["purchased"]
purchased_by_at_least_two = (
    (ali_purchases & sara_purchases)
    | (ali_purchases & ahmed_purchases)
    | (sara_purchases & ahmed_purchases)
)

ali_wishlist_minus_purchases = users["Ali"]["wishlist"] - users["Ali"]["purchased"]
shared_wishlist_products = (
    users["Ali"]["wishlist"] & users["Sara"]["wishlist"]
    & users["Ahmed"]["wishlist"]
)
all_products = (
    ali_purchases | sara_purchases | ahmed_purchases
    | users["Ali"]["wishlist"] | users["Sara"]["wishlist"] | users["Ahmed"]["wishlist"]
)

print("Products purchased by at least two users:", purchased_by_at_least_two)
print("Ali wishlist minus purchases:", ali_wishlist_minus_purchases)
print("Products on every user's wishlist:", shared_wishlist_products)
print("All products:", all_products)
```

## Question 50: Final Company Analysis

```python
# Given data
company = {
    "departments": {
        "Engineering": {
            "employees": [
                ("Ali", 120000, {"Python", "AWS", "SQL"}),
                ("Sara", 150000, {"Python", "Databricks", "SQL"})
            ],
            "budget": 500000
        },
        "Analytics": {
            "employees": [
                ("Ahmed", 100000, {"SQL", "Power BI"}),
                ("Zain", 130000, {"Python", "SQL", "Tableau"})
            ],
            "budget": 400000
        }
    }
}

# 1. Payroll per department.
engineering = company["departments"]["Engineering"]
analytics = company["departments"]["Analytics"]
engineering_payroll = engineering["employees"][0][1] + engineering["employees"][1][1]
analytics_payroll = analytics["employees"][0][1] + analytics["employees"][1][1]
payroll_per_department = {
    "Engineering": engineering_payroll,
    "Analytics": analytics_payroll
}

# 2. Company payroll.
company_payroll = engineering_payroll + analytics_payroll

# 3. Highest-paid employee.
name_to_salary = {
    engineering["employees"][0][0]: engineering["employees"][0][1],
    engineering["employees"][1][0]: engineering["employees"][1][1],
    analytics["employees"][0][0]: analytics["employees"][0][1],
    analytics["employees"][1][0]: analytics["employees"][1][1]
}
highest_paid_employee = max(name_to_salary, key=name_to_salary.get)

# 4. Skills shared by all employees.
ali_skills = engineering["employees"][0][2]
sara_skills = engineering["employees"][1][2]
ahmed_skills = analytics["employees"][0][2]
zain_skills = analytics["employees"][1][2]
skills_shared_by_all = ali_skills & sara_skills & ahmed_skills & zain_skills

# 5. Engineering-only skills (not used by Analytics employees).
engineering_skills = ali_skills | sara_skills
analytics_skills = ahmed_skills | zain_skills
engineering_only_skills = engineering_skills - analytics_skills

# 6. Remaining budget per department.
remaining_budget = {
    "Engineering": engineering["budget"] - engineering_payroll,
    "Analytics": analytics["budget"] - analytics_payroll
}

# 7. Department with higher average salary.
engineering_average_salary = engineering_payroll / len(engineering["employees"])
analytics_average_salary = analytics_payroll / len(analytics["employees"])
higher_average_salary_department = (
    "Engineering" if engineering_average_salary > analytics_average_salary else "Analytics"
)

# 8. Name-to-salary dictionary is name_to_salary above.

# 9. All unique skills.
all_unique_skills = ali_skills | sara_skills | ahmed_skills | zain_skills

# 10. Check whether combined budget is sufficient for company payroll.
combined_budget = engineering["budget"] + analytics["budget"]
budget_is_sufficient = combined_budget >= company_payroll

print("Payroll per department:", payroll_per_department)
print("Company payroll:", company_payroll)
print("Highest-paid employee:", highest_paid_employee, "with salary", name_to_salary[highest_paid_employee])
print("Skills shared by all employees:", skills_shared_by_all)
print("Engineering-only skills:", engineering_only_skills)
print("Remaining budget per department:", remaining_budget)
print("Average salary - Engineering:", engineering_average_salary)
print("Average salary - Analytics:", analytics_average_salary)
print("Department with higher average salary:", higher_average_salary_department)
print("Name-to-salary dictionary:", name_to_salary)
print("All unique skills:", all_unique_skills)
print("Combined budget:", combined_budget)
print("Is combined budget sufficient?", budget_is_sufficient)
```
