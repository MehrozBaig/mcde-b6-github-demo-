## Question 31: Dictionary Access

```python
# Given data
student = {"name": "Ali", "age": 22, "course": "Python", "marks": 85}

# Extract the requested values.
name = student["name"]
marks = student["marks"]
city = student.get("city", "Karachi")

print("Name:", name)
print("Marks:", marks)
print("City:", city)
```

## Question 32: Update Dictionary

```python
# Given data
employee = {"name": "Sara", "salary": 100000, "department": "IT"}

# Increase salary by 20%, add a bonus, and update the department.
employee["salary"] = employee["salary"] * 1.20
employee["bonus"] = 15000
employee["department"] = "Data Engineering"

print("Updated employee:", employee)
```

## Question 33: Dictionary Methods

```python
# Given data
product = {"id": 101, "name": "Laptop", "price": 120000, "stock": 5}

# Extract keys, values, and key-value pairs.
keys = product.keys()
values = product.values()
items = product.items()

# Remove stock while keeping the removed value.
stock = product.pop("stock")

print("Keys:", keys)
print("Values:", values)
print("Items:", items)
print("Removed stock value:", stock)
print("Product after removing stock:", product)
```

## Question 34: Merge Dictionaries

```python
# Given data
dict1 = {"a": 10, "b": 20, "c": 30}
dict2 = {"b": 50, "d": 40}

# Unpack both dictionaries; later values override earlier overlapping keys.
merged = {**dict1, **dict2}

print("Merged dictionary:", merged)
```

## Question 35: Student Marks

```python
# Given data
marks = {"Ali": 85, "Sara": 92, "Ahmed": 78, "Zain": 88}

# Calculate the highest, lowest, and average marks.
highest = max(marks.values())
lowest = min(marks.values())
average = sum(marks.values()) / len(marks)

# Find the student whose mark is the highest, without a loop or lambda.
top_student = max(marks, key=marks.get)

print("Highest marks:", highest)
print("Lowest marks:", lowest)
print("Average marks:", average)
print("Top-scoring student:", top_student)
```

## Question 36: Nested Employee Records

```python
# Given data
employees = {
    "E101": {"name": "Ali", "salary": 80000},
    "E102": {"name": "Sara", "salary": 120000},
    "E103": {"name": "Ahmed", "salary": 95000}
}

# Extract Sara's salary and increase Ali's salary by 15%.
sara_salary = employees["E102"]["salary"]
employees["E101"]["salary"] = employees["E101"]["salary"] * 1.15

# Calculate total payroll from the updated records.
total_payroll = (
    employees["E101"]["salary"]
    + employees["E102"]["salary"]
    + employees["E103"]["salary"]
)

print("Sara's salary:", sara_salary)
print("Ali's updated salary:", employees["E101"]["salary"])
print("Total payroll:", total_payroll)
```

## Question 37: Dictionary Inversion

```python
# Given data
data = {"Python": "Ali", "SQL": "Sara", "AWS": "Ahmed"}

# Swap keys and values using zip(), without loops or comprehensions.
inverse = dict(zip(data.values(), data.keys()))

print("Inverse dictionary:", inverse)
```

## Question 38: Order Calculation

```python
# Given data
order = {
    "laptop": {"price": 120000, "quantity": 2},
    "mouse": {"price": 1500, "quantity": 5},
    "keyboard": {"price": 3500, "quantity": 3}
}

# Calculate each subtotal from its price and quantity.
subtotals = {
    "laptop": order["laptop"]["price"] * order["laptop"]["quantity"],
    "mouse": order["mouse"]["price"] * order["mouse"]["quantity"],
    "keyboard": order["keyboard"]["price"] * order["keyboard"]["quantity"]
}

grand_total = sum(subtotals.values())
discount = grand_total * 0.10
amount_after_discount = grand_total - discount

print("Subtotals:", subtotals)
print("Grand total:", grand_total)
print("10% discount:", discount)
print("Amount after discount:", amount_after_discount)
```

## Question 39: Employee Performance

```python
# Given data
employees = {
    "Ali": {"python": 90, "sql": 85, "aws": 88},
    "Sara": {"python": 95, "sql": 92, "aws": 90},
    "Ahmed": {"python": 80, "sql": 87, "aws": 84}
}

# Calculate each employee's average score.
averages = {
    "Ali": (employees["Ali"]["python"] + employees["Ali"]["sql"] + employees["Ali"]["aws"]) / 3,
    "Sara": (employees["Sara"]["python"] + employees["Sara"]["sql"] + employees["Sara"]["aws"]) / 3,
    "Ahmed": (employees["Ahmed"]["python"] + employees["Ahmed"]["sql"] + employees["Ahmed"]["aws"]) / 3
}

# Find the top performer and highest SQL score without loops or lambda.
top_performer = max(averages, key=averages.get)
sql_scores = {
    "Ali": employees["Ali"]["sql"],
    "Sara": employees["Sara"]["sql"],
    "Ahmed": employees["Ahmed"]["sql"]
}
highest_sql_score = max(sql_scores.values())
highest_sql_employee = max(sql_scores, key=sql_scores.get)

print("Employee averages:", averages)
print("Top performer:", top_performer, "with average", averages[top_performer])
print("Highest SQL score:", highest_sql_score, "by", highest_sql_employee)
```

## Question 40: Inventory Management

```python
# Given data
inventory = {
    "P101": {"name": "Laptop", "price": 120000, "stock": 5},
    "P102": {"name": "Mouse", "price": 1500, "stock": 20},
    "P103": {"name": "Keyboard", "price": 3500, "stock": 10}
}

# Calculate total inventory value from price multiplied by stock.
item_values = {
    "P101": inventory["P101"]["price"] * inventory["P101"]["stock"],
    "P102": inventory["P102"]["price"] * inventory["P102"]["stock"],
    "P103": inventory["P103"]["price"] * inventory["P103"]["stock"]
}
total_value = sum(item_values.values())

# Identify the item with the greatest price-times-stock value.
most_valuable_id = max(item_values, key=item_values.get)
most_valuable_item = inventory[most_valuable_id]["name"]

# Update stock after selling two laptops and five mice.
inventory["P101"]["stock"] = inventory["P101"]["stock"] - 2
inventory["P102"]["stock"] = inventory["P102"]["stock"] - 5

print("Value by item:", item_values)
print("Total inventory value before sales:", total_value)
print("Most valuable item by price x stock:", most_valuable_item)
print("Updated inventory:", inventory)
print("Laptop stock after sale:", inventory["P101"]["stock"])
print("Mouse stock after sale:", inventory["P102"]["stock"])
```
